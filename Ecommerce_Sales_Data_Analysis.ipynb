# E-commerce Sales Data Analysis
## Sample Superstore Dataset

**Objective:** Explore sales, profit, customers, products, regions, discounts, and shipping performance to identify useful business insights.

**Dataset:** `Superstore.csv`  
**Tools:** Python, Pandas, NumPy, Matplotlib

> Keep `Superstore.csv` in the same folder as this notebook before running the cells.

## 1. Import Libraries

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

%matplotlib inline
```

## 2. Load Dataset

```python
print("Current working directory:", os.getcwd())

df = pd.read_csv("Superstore.csv", encoding="cp1252")
df.head()
```

## 3. Dataset Overview

```python
print("Dataset shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()
```

```python
display(df.head())
display(df.tail())
display(df.describe())
```

## 4. Data Cleaning and Validation

```python
print("Missing values in each column:")
display(df.isnull().sum())

print("Total missing values:", df.isnull().sum().sum())
print("Duplicate rows:", df.duplicated().sum())
```

```python
# Convert date columns to datetime format
df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Ship Date"] = pd.to_datetime(df["Ship Date"])

# Confirm data types after conversion
df[["Order Date", "Ship Date"]].dtypes
```

## 5. Category Analysis

```python
print("Categories:")
print(df["Category"].unique())

print("\nNumber of records in each category:")
display(df["Category"].value_counts())
```

```python
category_sales = df.groupby("Category")["Sales"].agg(
    Total_Sales="sum",
    Average_Sales="mean",
    Number_of_Records="count",
    Maximum_Sale="max",
    Minimum_Sale="min"
)
display(category_sales)
```

```python
category_profit = df.groupby("Category")["Profit"].agg(
    Total_Profit="sum",
    Average_Profit="mean"
)
display(category_profit)
```

```python
category_sales["Total_Sales"].sort_values(ascending=False).plot(
    kind="bar", figsize=(8, 5), title="Total Sales by Category"
)
plt.xlabel("Category")
plt.ylabel("Sales")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
category_profit["Total_Profit"].sort_values(ascending=False).plot(
    kind="bar", figsize=(8, 5), title="Total Profit by Category"
)
plt.xlabel("Category")
plt.ylabel("Profit")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## 6. Loss Analysis

```python
loss_records = df[df["Profit"] < 0]

print("Number of loss-making records:", len(loss_records))
print("Percentage of loss-making records:", round(len(loss_records) / len(df) * 100, 2), "%")

print("\nLoss-making records by category:")
display(loss_records["Category"].value_counts())

print("\nTotal negative profit by category:")
display(loss_records.groupby("Category")["Profit"].sum().sort_values())
```

## 7. Product Analysis

```python
product_sales = df.groupby("Product Name")["Sales"].sum().sort_values(ascending=False)
print("Top 5 products by sales:")
display(product_sales.head(5))
```

```python
product_profit = df.groupby("Product Name")["Profit"].sum().sort_values(ascending=False)

print("Top 5 products by profit:")
display(product_profit.head(5))

print("10 products with the lowest total profit:")
display(product_profit.head(10))
```

```python
product_sales.head(10).sort_values().plot(
    kind="barh", figsize=(10, 6), title="Top 10 Products by Sales"
)
plt.xlabel("Total Sales")
plt.ylabel("Product")
plt.tight_layout()
plt.show()
```

## 8. Regional and Segment Analysis

```python
regional_sales = df.groupby("Region")["Sales"].sum().sort_values(ascending=False)
regional_profit = df.groupby("Region")["Profit"].sum().sort_values(ascending=False)

print("Sales by region:")
display(regional_sales)

print("Profit by region:")
display(regional_profit)
```

```python
segment_sales = df.groupby("Segment")["Sales"].sum().sort_values(ascending=False)
segment_profit = df.groupby("Segment")["Profit"].sum().sort_values(ascending=False)

print("Sales by customer segment:")
display(segment_sales)

print("Profit by customer segment:")
display(segment_profit)
```

```python
regional_orders = df.groupby("Region")["Order ID"].nunique().sort_values(ascending=False)
print("Unique orders by region:")
display(regional_orders)
```

```python
regional_sales.plot(kind="bar", figsize=(8, 5), title="Sales by Region")
plt.xlabel("Region")
plt.ylabel("Sales")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## 9. Customer Analysis

```python
total_customers = df["Customer ID"].nunique()
customer_sales = df.groupby("Customer ID")["Sales"].sum().sort_values(ascending=False)
customer_profit = df.groupby("Customer ID")["Profit"].sum().sort_values(ascending=False)

print("Total unique customers:", total_customers)
print("Top 10 customers by sales:")
display(customer_sales.head(10))

print("Top 10 customers by profit:")
display(customer_profit.head(10))
```

```python
top_customer_sales = customer_sales.head(10)

top_customer_sales.sort_values().plot(
    kind="barh", figsize=(9, 6), title="Top 10 Customers by Sales"
)
plt.xlabel("Total Sales")
plt.ylabel("Customer ID")
plt.tight_layout()
plt.show()
```

## 10. City Analysis

```python
city_sales = df.groupby("City")["Sales"].sum().sort_values(ascending=False)
city_profit = df.groupby("City")["Profit"].sum().sort_values(ascending=False)

print("Top 10 cities by sales:")
display(city_sales.head(10))

print("Top 5 cities by profit:")
display(city_profit.head(5))
```

```python
city_sales.head(10).sort_values().plot(
    kind="barh", figsize=(9, 6), title="Top 10 Cities by Sales"
)
plt.xlabel("Total Sales")
plt.ylabel("City")
plt.tight_layout()
plt.show()
```

## 11. Region and Category Analysis

```python
region_category_sales = pd.pivot_table(
    df, values="Sales", index="Region", columns="Category", aggfunc="sum"
)
region_category_profit = pd.pivot_table(
    df, values="Profit", index="Region", columns="Category", aggfunc="sum"
)

print("Sales by region and category:")
display(region_category_sales)

print("Profit by region and category:")
display(region_category_profit)
```

```python
region_category_counts = pd.crosstab(df["Region"], df["Category"])
region_category_percent = pd.crosstab(
    df["Region"], df["Category"], normalize="index"
) * 100

print("Record counts by region and category:")
display(region_category_counts)

print("Percentage distribution within each region:")
display(region_category_percent.round(2))
```

```python
region_category_profit.plot(
    kind="bar", figsize=(10, 6), title="Profit by Region and Category"
)
plt.xlabel("Region")
plt.ylabel("Profit")
plt.xticks(rotation=0)
plt.legend(title="Category")
plt.tight_layout()
plt.show()
```

## 12. Correlation Analysis

```python
correlation = df[["Sales", "Profit", "Quantity", "Discount"]].corr()
display(correlation)
```

```python
plt.figure(figsize=(7, 5))
plt.imshow(correlation, cmap="coolwarm", interpolation="nearest")
plt.colorbar(label="Correlation")
plt.xticks(range(len(correlation.columns)), correlation.columns, rotation=45)
plt.yticks(range(len(correlation.index)), correlation.index)

for i in range(len(correlation.index)):
    for j in range(len(correlation.columns)):
        plt.text(j, i, f"{correlation.iloc[i, j]:.2f}",
                 ha="center", va="center")

plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```

## 13. Date and Monthly Sales Analysis

```python
df["Year"] = df["Order Date"].dt.year
df["Month"] = df["Order Date"].dt.month

monthly_sales = df.groupby("Month")["Sales"].sum()
display(monthly_sales)

print("Highest sales month:", monthly_sales.idxmax())
print("Lowest sales month:", monthly_sales.idxmin())
```

```python
monthly_sales.plot(
    kind="line", marker="o", figsize=(10, 5),
    title="Monthly Sales Trend (All Years Combined)"
)
plt.xlabel("Month Number")
plt.ylabel("Total Sales")
plt.xticks(range(1, 13))
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
yearly_sales = df.groupby("Year")["Sales"].sum()
yearly_profit = df.groupby("Year")["Profit"].sum()

print("Yearly sales:")
display(yearly_sales)

print("Yearly profit:")
display(yearly_profit)
```

```python
yearly_summary = pd.DataFrame({
    "Sales": yearly_sales,
    "Profit": yearly_profit
})
yearly_summary["Profit Margin (%)"] = (
    yearly_summary["Profit"] / yearly_summary["Sales"] * 100
)

display(yearly_summary)
```

```python
yearly_summary[["Sales", "Profit"]].plot(
    kind="line", marker="o", figsize=(9, 5),
    title="Yearly Sales and Profit"
)
plt.xlabel("Year")
plt.ylabel("Amount")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
yearly_summary["Profit Margin (%)"].plot(
    kind="line", marker="o", figsize=(9, 5),
    title="Yearly Profit Margin"
)
plt.xlabel("Year")
plt.ylabel("Profit Margin (%)")
plt.grid(True)
plt.tight_layout()
plt.show()
```

## 14. Shipping Analysis

```python
df["Shipping Days"] = (df["Ship Date"] - df["Order Date"]).dt.days

average_shipping_days = df.groupby("Ship Mode")["Shipping Days"].mean().sort_values()
display(average_shipping_days)
```

```python
shipping_performance = df.groupby("Ship Mode").agg(
    Sales=("Sales", "sum"),
    Profit=("Profit", "sum"),
    Average_Shipping_Days=("Shipping Days", "mean")
)
display(shipping_performance)
```

```python
average_shipping_days.plot(
    kind="bar", figsize=(8, 5), title="Average Shipping Days by Ship Mode"
)
plt.xlabel("Ship Mode")
plt.ylabel("Average Shipping Days")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

```python
shipping_performance[["Sales", "Profit"]].plot(
    kind="bar", figsize=(9, 5), title="Sales and Profit by Ship Mode"
)
plt.xlabel("Ship Mode")
plt.ylabel("Amount")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

**Interpretation note:** The dataset stores order and ship dates without exact times. Therefore, a very small average for Same Day shipping reflects date-level measurement and should not be interpreted as a precise delivery duration in hours.

## 15. Repeat Customer Analysis

```python
customer_orders = df.groupby("Customer ID")["Order ID"].nunique()

repeat_customers = (customer_orders > 1).sum()
single_order_customers = (customer_orders == 1).sum()
repeat_customer_percentage = repeat_customers / len(customer_orders) * 100

print("Total customers:", len(customer_orders))
print("Repeat customers:", repeat_customers)
print("Single-order customers:", single_order_customers)
print("Repeat customer percentage:", round(repeat_customer_percentage, 2), "%")
```

**Interpretation note:** This percentage describes repeat ordering within the available dataset period. It does not, by itself, prove long-term customer loyalty.

## 16. Customer Sales vs Profit Analysis

```python
customer_comparison = df.groupby("Customer ID").agg(
    Sales=("Sales", "sum"),
    Profit=("Profit", "sum")
)

top_10_sales_customers = customer_comparison.sort_values(
    "Sales", ascending=False
).head(10)

display(top_10_sales_customers)
```

```python
top_10_sales_customers.plot(
    kind="bar", figsize=(11, 6), title="Top 10 Customers: Sales vs Profit"
)
plt.xlabel("Customer ID")
plt.ylabel("Amount")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
customer_comparison["Profit Margin"] = (
    customer_comparison["Profit"] / customer_comparison["Sales"] * 100
)

top_10_sales_customers = customer_comparison.sort_values(
    "Sales", ascending=False
).head(10)

display(top_10_sales_customers)
```

```python
top_10_sales_customers["Profit Margin"].sort_values().plot(
    kind="barh", figsize=(9, 6), title="Profit Margin of Top 10 Sales Customers"
)
plt.xlabel("Profit Margin (%)")
plt.ylabel("Customer ID")
plt.tight_layout()
plt.show()
```

## 17. Discount Analysis

```python
discount_profit = df.groupby("Discount")["Profit"].mean().sort_index()
discount_sales = df.groupby("Discount")["Sales"].mean().sort_index()
discount_quantity = df.groupby("Discount")["Quantity"].sum().sort_index()

print("Average profit by discount:")
display(discount_profit)

print("Average sales by discount:")
display(discount_sales)

print("Total quantity by discount:")
display(discount_quantity)
```

```python
discount_profit.plot(
    kind="line", marker="o", figsize=(9, 5),
    title="Discount vs Average Profit"
)
plt.xlabel("Discount")
plt.ylabel("Average Profit")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
print("Discount level with highest average profit:", discount_profit.idxmax())
print("Discount level with lowest average profit:", discount_profit.idxmin())
```

**Interpretation note:** These results show associations in this dataset. They do not prove that discount level alone caused a change in profit; product mix, order size, and other factors may also matter.

## 18. KPI Summary

```python
total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()
total_orders = df["Order ID"].nunique()
total_customers = df["Customer ID"].nunique()
overall_profit_margin = total_profit / total_sales * 100

kpi_summary = pd.Series({
    "Total Sales": total_sales,
    "Total Profit": total_profit,
    "Total Orders": total_orders,
    "Total Customers": total_customers,
    "Overall Profit Margin (%)": overall_profit_margin
})

display(kpi_summary)
```

## 19. Sales Analysis Dashboard

```python
fig, axes = plt.subplots(2, 2, figsize=(15, 10))

df.groupby("Category")["Sales"].sum().plot(
    kind="bar", ax=axes[0, 0], title="Sales by Category"
)
axes[0, 0].set_xlabel("Category")
axes[0, 0].set_ylabel("Sales")
axes[0, 0].tick_params(axis="x", rotation=0)

df.groupby("Region")["Profit"].sum().plot(
    kind="bar", ax=axes[0, 1], title="Profit by Region"
)
axes[0, 1].set_xlabel("Region")
axes[0, 1].set_ylabel("Profit")
axes[0, 1].tick_params(axis="x", rotation=0)

df.groupby("Month")["Sales"].sum().plot(
    kind="line", marker="o", ax=axes[1, 0], title="Monthly Sales Trend"
)
axes[1, 0].set_xlabel("Month")
axes[1, 0].set_ylabel("Sales")
axes[1, 0].set_xticks(range(1, 13))
axes[1, 0].grid(True)

df.groupby("Category")[["Sales", "Profit"]].sum().plot(
    kind="bar", ax=axes[1, 1], title="Sales vs Profit by Category"
)
axes[1, 1].set_xlabel("Category")
axes[1, 1].set_ylabel("Amount")
axes[1, 1].tick_params(axis="x", rotation=0)

plt.tight_layout()
plt.show()
```

## 20. Key Business Insights

```python
print("1. Highest sales category:", category_sales["Total_Sales"].idxmax())
print("2. Highest profit category:", category_profit["Total_Profit"].idxmax())
print("3. Region with highest sales:", regional_sales.idxmax())
print("4. Region with highest profit:", regional_profit.idxmax())
print("5. Discount with highest average profit:", discount_profit.idxmax())
print("6. Highest sales month (all years combined):", monthly_sales.idxmax())
print("7. Year with highest sales:", yearly_sales.idxmax())
print("8. Year with highest profit:", yearly_profit.idxmax())
```

### Findings to discuss

- Technology generated the highest total sales and total profit among the three categories.
- The West region recorded the highest total sales and profit.
- The Consumer segment contributed the largest sales and profit totals.
- November had the highest combined monthly sales; February had the lowest.
- 2017 had the highest yearly sales and profit in the dataset.
- Some products and customer transactions generated negative profit, so sales volume should be considered alongside profitability.
- Higher discount levels were often associated with lower average profit in this dataset. This is an association, not proof of causation.
- The top-selling customer was not necessarily the most profitable customer.
